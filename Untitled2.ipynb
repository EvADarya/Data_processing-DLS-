```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder, LabelEncoder, StandardScaler
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import numpy as np
```

```python
data['alcohol_freq'] = data['alcohol_freq'].fillna('Occasional')
```

```python
object_data = pd.DataFrame(pd.concat([data.plan_type, data.urban_rural, data.region, data.education,data.marital_status, data.employment_status, data.smoker,data.alcohol_freq, data.network_tier], axis = 1))
```

```python
object_data_encoded = object_data.copy()
for col in object_data.columns:
    le = LabelEncoder()
    object_data_encoded[col] = le.fit_transform(object_data[col])
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    full_data,
    target_col,
    test_size=0.2,
    random_state=42,
    shuffle=True
)
```

```python
X_test.dropna(inplace= True)
```

```python
X_train.dropna(inplace= True)
```

```python
y_train.dropna()
```

```python
X_train
```

```python

models = {
    'Линейная регрессия': LinearRegression(),
    'Ridge (α=1.0)': Ridge(alpha=1.0, random_state=42),
    'LASSO (α=0.01)': Lasso(alpha=0.01, random_state=42, max_iter=10000)
}


results = {}
predictions = {}


for name, model in models.items():
    print(f"Обучение модели: {name}")
    model.fit(X_train, y_train)

    y_pred_train = model.predict(X_train)
    y_pred_test = model.predict(X_test)


    predictions[name] = {
        'train': y_pred_train,
        'test': y_pred_test
    }


    train_rmse = np.sqrt(mean_squared_error(y_train, y_pred_train))
    test_rmse = np.sqrt(mean_squared_error(y_test, y_pred_test))
    train_mae = mean_absolute_error(y_train, y_pred_train)
    test_mae = mean_absolute_error(y_test, y_pred_test)
    train_r2 = r2_score(y_train, y_pred_train)
    test_r2 = r2_score(y_test, y_pred_test)

    results[name] = {
        'train_rmse': train_rmse,
        'test_rmse': test_rmse,
        'train_mae': train_mae,
        'test_mae': test_mae,
        'train_r2': train_r2,
        'test_r2': test_r2,
        'model': model
    }

    print(f"Train RMSE: {train_rmse:.2f}")
    print(f"Test RMSE:  {test_rmse:.2f}")
    print(f"Train MAE:  {train_mae:.2f}")
    print(f"Test MAE:   {test_mae:.2f}")
    print(f"Train R²:   {train_r2:.4f}")
    print(f"Test R²:    {test_r2:.4f}")
```

```python
import matplotlib.pyplot as plt
import numpy as np

colors = ['blue', 'green', 'red']
model_names = list(results.keys())

fig, axes = plt.subplots(1, 3, figsize=(18, 6))

#Linear regression
ax = axes[0]
name = 'Линейная регрессия'
y_pred_test = predictions[name]['test']

ax.scatter(y_test, y_pred_test, alpha=0.7, s=50, color='blue')

min_val = min(y_test.min(), y_pred_test.min())
max_val = max(y_test.max(), y_pred_test.max())
ax.plot([min_val, max_val], [min_val, max_val], 'k--', alpha=0.7, linewidth=2)

ax.set_xlabel('Фактические значения', fontsize=11)
ax.set_ylabel('Предсказанные значения', fontsize=11)
ax.set_title(f'{name}\nTest R² = {results[name]["test_r2"]:.4f}', fontsize=13)
ax.grid(True, alpha=0.3)

# 2 Ridge
ax = axes[1]
name = 'Ridge (α=1.0)'
y_pred_test = predictions[name]['test']

ax.scatter(y_test, y_pred_test, alpha=0.7, s=50, color='green')

min_val = min(y_test.min(), y_pred_test.min())
max_val = max(y_test.max(), y_pred_test.max())
ax.plot([min_val, max_val], [min_val, max_val], 'k--', alpha=0.7, linewidth=2)

ax.set_xlabel('Фактические значения', fontsize=11)
ax.set_title(f'{name}\nTest R² = {results[name]["test_r2"]:.4f}', fontsize=13)
ax.grid(True, alpha=0.3)

# 3.LASSO
ax = axes[2]
name = 'LASSO (α=0.01)'
y_pred_test = predictions[name]['test']

ax.scatter(y_test, y_pred_test, alpha=0.7, s=50, color='red')

# Линия идеального предсказания
min_val = min(y_test.min(), y_pred_test.min())
max_val = max(y_test.max(), y_pred_test.max())
ax.plot([min_val, max_val], [min_val, max_val], 'k--', alpha=0.7, linewidth=2)

ax.set_xlabel('Фактические значения', fontsize=11)
ax.set_title(f'{name}\nTest R² = {results[name]["test_r2"]:.4f}', fontsize=13)
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
